#### 17. What is the result of the following expression? (★☆☆)
```python
0 * np.nan
np.nan == np.nan
np.inf > np.nan
np.nan - np.nan
np.nan in set([np.nan])
0.3 == 3 * 0.1
```

````python
0 * np.nan
````

````python
np.nan == np.nan
````

因为`np.nan`（Not a Number）表示一个未定义或不可表示的值，它不与任何值相等，包括它自己。需要使用`isnan`判断

````python
1 if np.isnan(np.nan) else 0
````

````python
1 if np.isnan(3) else 0
````

````python
np.inf > np.nan
````

````python
np.nan - np.nan # 对np.nan的任何运算都是nan
````

````python
np.nan in set([np.nan])
````

````python
print(hash(np.nan))  # 输出一个固定的整数值
print(hash(float('nan')))
````

````python
print(np.nan is np.nan)  # 输出: True
````

````python
s = set([np.nan, float('nan'), np.nan])
print(s)  # np.nan和float('nan')是不一样的
````

````python
0.3 == 3 * 0.1
````

#### 23. Create a custom dtype that describes a color as four unsigned bytes (RGBA) (★☆☆)

````python
hint(23)
answer(23)
````

````python
color = np.dtype([
    ("r", np.ubyte),
    ("g", np.ubyte),
    ("b", np.ubyte),
    ("a", np.ubyte),
])
print(color)
````

````python
colors = np.array([(255, 0, 0, 255), (0, 255, 0, 255), (0, 0, 255, 255)], dtype=color)
print(colors)
````

````python
print(colors["r"]) # 访问所有元素的红色通道
````

````python
# 访问第一个元素的绿色通道
print(colors[0]["g"])  # 输出: 0
````

````python
# 修改第二个元素的蓝色通道
colors[1]["b"] = 128
print(colors)
````

````python
for c in colors: # 遍历数组中的每个元素，并访问其字段。
    print(f"r={c['r']}, g={c['g']}, b={c['b']}, a={c['a']}")
````
